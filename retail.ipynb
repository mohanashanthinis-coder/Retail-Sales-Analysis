```python
sales['order_date'] = pd.to_datetime(sales['order_date'],format='%d-%m-%Y')
```

```python
sales['order_date'] 
```

```python
reference_date=sales['order_date'].max()+pd.Timedelta(days=1)
```

```python
reference_date
```

```python
newtable=sales.groupby('customer_id').agg({
    'order_date':lambda x:(reference_date-x.max()).days,
    'order_id':'nunique',
    'total_price':'sum'})
newtable.columns=['recency','frequency','monetary']
newtable=newtable.reset_index()
newtable
```

```python
newtable['segment'] = 'Regular Customer'

newtable.loc[
    (newtable['recency'] <= 635) &
    (newtable['frequency'] >= 2) &
    (newtable['monetary'] >= 4361.66),
    'segment'
] = 'Premium Customer'
```

```python
newtable.loc[
    (newtable['recency'] > 635) &
    (newtable['frequency'] == 1),
    'segment'
] = 'At Risk Customer'
newtable
```

```python
from sqlalchemy import create_engine
engine = create_engine("mysql+pymysql://root:12345@localhost:3306/retail")
newtable.to_sql(
    name='customer_segments',
    con=engine,
    if_exists='replace',
    index=False
)

print("Customer Segments exported successfully!")
```
